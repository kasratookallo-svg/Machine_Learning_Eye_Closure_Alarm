```python
import pandas as pd
import joblib

from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)


# ============================================================
# CONFIGURATION
# ============================================================

FEATURES = [
    "left_ear",
    "right_ear",
    "ear",
    "left_eye_width",
    "left_eye_height_1",
    "left_eye_height_2",
    "right_eye_width",
    "right_eye_height_1",
    "right_eye_height_2"
]

TARGET = "label"

TEST_SIZE = 0.20
RANDOM_STATE = 42


# ============================================================
# FUNCTION
# ============================================================

def evaluate_model(dataset_file, model_file, model_name):

    print()
    print("=" * 70)
    print(model_name)
    print("=" * 70)

    df = pd.read_csv(dataset_file)

    model = joblib.load(model_file)

    X = df[FEATURES]
    y = df[TARGET]

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=TEST_SIZE,
        random_state=RANDOM_STATE,
        stratify=y
    )

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(
        y_test,
        y_pred
    )

    precision = precision_score(
        y_test,
        y_pred,
        pos_label="CLOSED"
    )

    recall = recall_score(
        y_test,
        y_pred,
        pos_label="CLOSED"
    )

    f1 = f1_score(
        y_test,
        y_pred,
        pos_label="CLOSED"
    )

    print()
    print(f"Dataset samples : {len(df)}")
    print(f"Test samples    : {len(X_test)}")

    print()
    print(f"Accuracy          : {accuracy:.4f}")
    print(f"Precision CLOSED  : {precision:.4f}")
    print(f"Recall CLOSED     : {recall:.4f}")
    print(f"F1 CLOSED         : {f1:.4f}")

    return {
        "Model": model_name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    }


# ============================================================
# MODEL 1
# ============================================================

result_1 = evaluate_model(
    "eye_dataset_v2.csv",
    "eye_random_forest_v2.pkl",
    "MODEL 1 - ORIGINAL DATASET"
)


# ============================================================
# MODEL 2
# ============================================================

result_2 = evaluate_model(
    "eye_dataset_clean.csv",
    "eye_random_forest_clean.pkl",
    "MODEL 2 - CLEAN DATASET"
)


# ============================================================
# COMPARISON
# ============================================================

results = pd.DataFrame([
    result_1,
    result_2
])


print()
print("=" * 70)
print("MODEL COMPARISON")
print("=" * 70)

print()
print(
    results.to_string(index=False)
)


# ============================================================
# BEST MODEL
# ============================================================

best = results.loc[
    results["F1"].idxmax()
]

print()
print("=" * 70)
print("BEST MODEL")
print("=" * 70)

print(
    f"Model : {best['Model']}"
)

print(
    f"F1    : {best['F1']:.4f}"
)

print()
```
