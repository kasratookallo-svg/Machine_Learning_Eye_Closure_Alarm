# Import Libraries

```python
import pandas as pd
import joblib

from sklearn.ensemble import RandomForestClassifier
```

```python

DATASET_FILE = "eye_dataset_clean.csv"
MODEL_FILE = "eye_random_forest_clean.pkl"

FEATURES = [
    "left_ear",
    "right_ear",
    "ear",
    "left_eye_width",
    "left_eye_height_1",
    "left_eye_height_2",
    "right_eye_width",
    "right_eye_height_1",
    "right_eye_height_2"
]

TARGET = "label"
```

## Train RANDOM FOREST

```python

print()
print("=" * 70)
print("TRAINING RANDOM FOREST MODEL")
print("=" * 70)

print()
print("Loading dataset...")

df = pd.read_csv(DATASET_FILE)


X = df[FEATURES]
y = df[TARGET]

print()
print("DATASET")
print("-" * 70)

print(f"Samples  : {len(df)}")
print(f"Features : {len(FEATURES)}")

print()
print("Classes:")
print(y.value_counts())



print()
print("=" * 70)
print("TRAINING")
print("=" * 70)

model = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1
)

model.fit(X, y)



# SAVE MODEL


joblib.dump(
    model,
    MODEL_FILE
)


# RESULT


print()
print("=" * 70)
print("MODEL TRAINING COMPLETE")
print("=" * 70)

print()
print("Model saved as:")
print(MODEL_FILE)

print()
print("Classes learned by model:")
print(model.classes_)

print()
print("Features used by model:")

for i, feature in enumerate(FEATURES, 1):
    print(f"{i}. {feature}")

print()
```
