```python
import pandas as pd
import numpy as np

DATASET_FILE = "eye_dataset_v2.csv"

df = pd.read_csv(DATASET_FILE)


# ============================================================
# Dataset Information
# ============================================================

print("=" * 60)
print("DATASET INFORMATION")
print("=" * 60)

print()
print("Dataset shape:")
print(df.shape)

print()
print("Dataset columns:")
print(df.columns.tolist())

print()
print("Class distribution:")
print(df["label"].value_counts())

print()
print("Missing values:")
print(df.isnull().sum())


FEATURES = [
    "left_ear",
    "right_ear",
    "ear",

    "left_eye_width",
    "left_eye_height_1",
    "left_eye_height_2",

    "right_eye_width",
    "right_eye_height_1",
    "right_eye_height_2"
]


# Feature Statistics by Class

print()
print("=" * 60)
print("FEATURE STATISTICS BY CLASS")
print("=" * 60)


for feature in FEATURES:

    print()
    print("-" * 60)
    print(f"Feature: {feature}")
    print("-" * 60)

    statistics = (
        df.groupby("label")[feature]
        .agg([
            "count",
            "mean",
            "std",
            "min",
            "max"
        ])
    )

    print(statistics)



# Median and Quartiles

print()
print("=" * 60)
print("MEDIAN AND QUARTILES")
print("=" * 60)

for feature in FEATURES:

    print()
    print("-" * 60)
    print(f"Feature: {feature}")
    print("-" * 60)

    statistics = (
        df.groupby("label")[feature]
        .describe()
    )

    print(
        statistics[
            [
                "count",
                "mean",
                "std",
                "min",
                "25%",
                "50%",
                "75%",
                "max"
            ]
        ]
    )



# Feature Correlation


print()
print("=" * 60)
print("FEATURE CORRELATION")
print("=" * 60)

correlation_matrix = df[FEATURES].corr()

print()
print(correlation_matrix)


# Correlation With EAR


print()
print("=" * 60)
print("CORRELATION WITH EAR")
print("=" * 60)

ear_correlation = (
    df[FEATURES]
    .corr()["ear"]
    .sort_values(ascending=False)
)

print()
print(ear_correlation)



print()
print("=" * 60)
print("EDA COMPLETE")
print("=" * 60)
```
